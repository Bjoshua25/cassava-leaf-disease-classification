```python
# ===== DOWNLOAD DATASET INTO GOOGLE DRIVE ======
import os
import zipfile
import requests
from tqdm import tqdm



# define google Drive path
PROJECT_ROOT = "/content/drive/MyDrive/cassava_leaf_disease_classification"
raw_data_dir = os.path.join(PROJECT_ROOT, "data/raw")

# verify if folder exist or create
os.makedirs(raw_data_dir, exist_ok=True)

# Setup Kaggle API authentication for python requests
with open("/root/.kaggle/access_token", "r") as f:
    token = f.read().strip()

# the request configuration for download
headers = {"Authorization": f"Bearer {token}"}
# Direct API download URL for the Cassava dataset zip file
url = "https://www.kaggle.com/api/v1/competitions/data/download-all/cassava-leaf-disease-classification"

# Zip file name and directory
zip_filename = "cassava-leaf-disease-classification.zip"
zip_path = os.path.join(raw_data_dir, zip_filename)

# Download with a tqdm progress bar
print("Downloading Cassava dataset...")
response = requests.get(url, headers=headers, stream=True)
total_size = int(response.headers.get('content-length', 0))

# Write chunk by chunk with a progress bar
with open(zip_path, "wb") as f, tqdm(
    desc="Downloading",
    total=total_size,
    unit='B',
    unit_scale=True,
    unit_divisor=1024,
) as bar:
    for data in response.iter_content(chunk_size=1024):
        f.write(data)
        bar.update(len(data))



print(f"\nDataset successfully downloaded to: {raw_data_dir}")
```

```python
# ==== UNZIP AND EXTRACT TO RAW DATA FOLDER ======



print("\nExtracting files to Google Drive...")

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    # Extract file list with a progress bar
    members = zip_ref.infolist()
    for member in tqdm(members, desc="Extracting", unit="file"):
        zip_ref.extract(member, raw_data_dir)

print(f"\nDataset successfully extracted to: {raw_data_dir}")
```
